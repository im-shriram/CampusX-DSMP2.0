I'll derive the closed-form OLS solution for Multiple Linear Regression step by step, explaining every mathematical operation clearly.

---

## **The Problem Setup**

We want to predict a target variable **y** using **p** predictor variables. For **n** observations:

$$\mathbf{y} = \mathbf{X}\boldsymbol{\beta} + \boldsymbol{\varepsilon}$$

Where:
- $\mathbf{y}$ is $n \times 1$ (target vector)
- $\mathbf{X}$ is $n \times (p+1)$ **design matrix** (includes a column of 1s for the intercept)
- $\boldsymbol{\beta}$ is $(p+1) \times 1$ (coefficient vector we want to find)
- $\boldsymbol{\varepsilon}$ is $n \times 1$ (error vector)

---

## **Step 1: Define the Objective (Residual Sum of Squares)**

We want to minimize the **sum of squared residuals**. For a single prediction:

$$\hat{y}_i = \beta_0 + \beta_1 x_{i1} + \beta_2 x_{i2} + ... + \beta_p x_{ip}$$

The residual for observation $i$:
$$e_i = y_i - \hat{y}_i = y_i - \mathbf{x}_i^T\boldsymbol{\beta}$$

**Residual Sum of Squares (RSS)** in matrix form:
$$\text{RSS}(\boldsymbol{\beta}) = \sum_{i=1}^{n} e_i^2 = \mathbf{e}^T\mathbf{e} = (\mathbf{y} - \mathbf{X}\boldsymbol{\beta})^T(\mathbf{y} - \mathbf{X}\boldsymbol{\beta})$$

---

## **Step 2: Expand the RSS Expression**

Let's expand this carefully using matrix transpose properties $(\mathbf{AB})^T = \mathbf{B}^T\mathbf{A}^T$:

$$\text{RSS}(\boldsymbol{\beta}) = (\mathbf{y} - \mathbf{X}\boldsymbol{\beta})^T(\mathbf{y} - \mathbf{X}\boldsymbol{\beta})$$

$$= \mathbf{y}^T\mathbf{y} - \mathbf{y}^T\mathbf{X}\boldsymbol{\beta} - (\mathbf{X}\boldsymbol{\beta})^T\mathbf{y} + (\mathbf{X}\boldsymbol{\beta})^T(\mathbf{X}\boldsymbol{\beta})$$

$$= \mathbf{y}^T\mathbf{y} - \mathbf{y}^T\mathbf{X}\boldsymbol{\beta} - \boldsymbol{\beta}^T\mathbf{X}^T\mathbf{y} + \boldsymbol{\beta}^T\mathbf{X}^T\mathbf{X}\boldsymbol{\beta}$$

Notice that $\mathbf{y}^T\mathbf{X}\boldsymbol{\beta}$ is a **scalar** (1×1), so it equals its own transpose:
$$\mathbf{y}^T\mathbf{X}\boldsymbol{\beta} = (\mathbf{y}^T\mathbf{X}\boldsymbol{\beta})^T = \boldsymbol{\beta}^T\mathbf{X}^T\mathbf{y}$$

So we can combine the middle terms:
$$\text{RSS}(\boldsymbol{\beta}) = \mathbf{y}^T\mathbf{y} - 2\boldsymbol{\beta}^T\mathbf{X}^T\mathbf{y} + \boldsymbol{\beta}^T\mathbf{X}^T\mathbf{X}\boldsymbol{\beta}$$

---

## **Step 3: Take the Derivative with Respect to β**

This is the **critical step**. We need to differentiate RSS with respect to the vector $\boldsymbol{\beta}$.

### **Key Matrix Calculus Rules:**

| Expression | Derivative |
|------------|-----------|
| $\mathbf{a}^T\mathbf{x}$ | $\mathbf{a}$ |
| $\mathbf{x}^T\mathbf{A}\mathbf{x}$ | $2\mathbf{A}\mathbf{x}$ (when **A** is symmetric) |

Applying these to each term in RSS:

**Term 1:** $\frac{\partial}{\partial\boldsymbol{\beta}}(\mathbf{y}^T\mathbf{y}) = \mathbf{0}$ *(constant, no β)*

**Term 2:** $\frac{\partial}{\partial\boldsymbol{\beta}}(-2\boldsymbol{\beta}^T\mathbf{X}^T\mathbf{y}) = -2\mathbf{X}^T\mathbf{y}$

**Term 3:** $\frac{\partial}{\partial\boldsymbol{\beta}}(\boldsymbol{\beta}^T\mathbf{X}^T\mathbf{X}\boldsymbol{\beta}) = 2\mathbf{X}^T\mathbf{X}\boldsymbol{\beta}$

*(Note: $\mathbf{X}^T\mathbf{X}$ is always symmetric)*

---

## **Step 4: Set Derivative to Zero (First-Order Condition)**

$$\frac{\partial \text{RSS}}{\partial \boldsymbol{\beta}} = -2\mathbf{X}^T\mathbf{y} + 2\mathbf{X}^T\mathbf{X}\boldsymbol{\beta} = \mathbf{0}$$

Divide by 2:
$$-\mathbf{X}^T\mathbf{y} + \mathbf{X}^T\mathbf{X}\boldsymbol{\beta} = \mathbf{0}$$

Rearrange:
$$\mathbf{X}^T\mathbf{X}\boldsymbol{\beta} = \mathbf{X}^T\mathbf{y}$$

---

## **Step 5: Solve for β (The Normal Equations)**

These are called the **Normal Equations**:

$$\boxed{\boldsymbol{\hat{\beta}} = (\mathbf{X}^T\mathbf{X})^{-1}\mathbf{X}^T\mathbf{y}}$$

This is the **closed-form OLS estimator**!

---

## **Why This Works: The Geometry**

The solution projects **y** onto the column space of **X**:

$$\hat{\mathbf{y}} = \mathbf{X}\boldsymbol{\hat{\beta}} = \mathbf{X}(\mathbf{X}^T\mathbf{X})^{-1}\mathbf{X}^T\mathbf{y} = \mathbf{H}\mathbf{y}$$

Where $\mathbf{H} = \mathbf{X}(\mathbf{X}^T\mathbf{X})^{-1}\mathbf{X}^T$ is the **"Hat Matrix"** — it "puts the hat on y."

The residual vector $\mathbf{e} = \mathbf{y} - \hat{\mathbf{y}}$ is **orthogonal** to the column space of **X**, which is why $\mathbf{X}^T\mathbf{e} = \mathbf{0}$ (this is where "normal" comes from — perpendicular).

---

## **Step 6: Verify It's a Minimum (Second-Order Condition)**

Take the second derivative (Hessian matrix):
$$\frac{\partial^2 \text{RSS}}{\partial \boldsymbol{\beta}\partial\boldsymbol{\beta}^T} = 2\mathbf{X}^T\mathbf{X}$$

For this to be a minimum, the Hessian must be **positive definite**.

- $\mathbf{X}^T\mathbf{X}$ is always **positive semi-definite**
- It's **positive definite** (and thus invertible) when **X** has full column rank (no perfect multicollinearity)
- This guarantees RSS is convex and our critical point is the **global minimum**

---

## **Complete Formula Summary**

| Component | Formula | Dimension |
|-----------|---------|-----------|
| **OLS Estimator** | $\boldsymbol{\hat{\beta}} = (\mathbf{X}^T\mathbf{X})^{-1}\mathbf{X}^T\mathbf{y}$ | $(p+1) \times 1$ |
| **Fitted Values** | $\hat{\mathbf{y}} = \mathbf{X}\boldsymbol{\hat{\beta}} = \mathbf{H}\mathbf{y}$ | $n \times 1$ |
| **Residuals** | $\mathbf{e} = \mathbf{y} - \hat{\mathbf{y}}$ | $n \times 1$ |
| **Hat Matrix** | $\mathbf{H} = \mathbf{X}(\mathbf{X}^T\mathbf{X})^{-1}\mathbf{X}^T$ | $n \times n$ |
| **RSS** | $\mathbf{e}^T\mathbf{e} = \mathbf{y}^T\mathbf{y} - \boldsymbol{\hat{\beta}}^T\mathbf{X}^T\mathbf{y}$ | scalar |

---

## **Key Metrics & Their Derivations**

### **1. Variance of the Estimator**
$$\text{Var}(\boldsymbol{\hat{\beta}}) = \sigma^2(\mathbf{X}^T\mathbf{X})^{-1}$$

*Derivation:* Assuming $\text{Var}(\mathbf{y}) = \sigma^2\mathbf{I}$:
$$\text{Var}(\boldsymbol{\hat{\beta}}) = \text{Var}((\mathbf{X}^T\mathbf{X})^{-1}\mathbf{X}^T\mathbf{y}) = (\mathbf{X}^T\mathbf{X})^{-1}\mathbf{X}^T \cdot \sigma^2\mathbf{I} \cdot \mathbf{X}(\mathbf{X}^T\mathbf{X})^{-1} = \sigma^2(\mathbf{X}^T\mathbf{X})^{-1}$$

### **2. Residual Standard Error (RSE)**
$$\hat{\sigma}^2 = \frac{\text{RSS}}{n - p - 1} = \frac{\mathbf{e}^T\mathbf{e}}{n - p - 1}$$

The denominator $n-p-1$ is **degrees of freedom** — we lose one for each parameter estimated (including intercept).

### **3. Standard Errors of Coefficients**
$$\text{SE}(\hat{\beta}_j) = \sqrt{\hat{\sigma}^2 [(\mathbf{X}^T\mathbf{X})^{-1}]_{jj}}$$

### **4. t-Statistic for Hypothesis Testing**
$$t_j = \frac{\hat{\beta}_j}{\text{SE}(\hat{\beta}_j)} \sim t_{n-p-1}$$

Tests $H_0: \beta_j = 0$ vs $H_1: \beta_j \neq 0$

### **5. R-squared (Coefficient of Determination)**
$$R^2 = 1 - \frac{\text{RSS}}{\text{TSS}} = 1 - \frac{\sum e_i^2}{\sum(y_i - \bar{y})^2}$$

Measures proportion of variance in **y** explained by the model. For multiple regression, use **adjusted R²**:
$$R^2_{adj} = 1 - \frac{\text{RSS}/(n-p-1)}{\text{TSS}/(n-1)}$$

## **Simple Numerical Example**

Suppose:
$$\mathbf{X} = \begin{bmatrix} 1 & 2 \\ 1 & 3 \\ 1 & 4 \end{bmatrix}, \quad \mathbf{y} = \begin{bmatrix} 5 \\ 7 \\ 9 \end{bmatrix}$$

**Step A:** Compute $\mathbf{X}^T\mathbf{X}$:
$$\mathbf{X}^T\mathbf{X} = \begin{bmatrix} 1 & 1 & 1 \\ 2 & 3 & 4 \end{bmatrix}\begin{bmatrix} 1 & 2 \\ 1 & 3 \\ 1 & 4 \end{bmatrix} = \begin{bmatrix} 3 & 9 \\ 9 & 29 \end{bmatrix}$$

**Step B:** Compute $(\mathbf{X}^T\mathbf{X})^{-1}$:
$$\det = 3(29) - 9(9) = 87 - 81 = 6$$
$$(\mathbf{X}^T\mathbf{X})^{-1} = \frac{1}{6}\begin{bmatrix} 29 & -9 \\ -9 & 3 \end{bmatrix}$$

**Step C:** Compute $\mathbf{X}^T\mathbf{y}$:
$$\mathbf{X}^T\mathbf{y} = \begin{bmatrix} 1 & 1 & 1 \\ 2 & 3 & 4 \end{bmatrix}\begin{bmatrix} 5 \\ 7 \\ 9 \end{bmatrix} = \begin{bmatrix} 21 \\ 67 \end{bmatrix}$$

**Step D:** Final calculation:
$$\boldsymbol{\hat{\beta}} = \frac{1}{6}\begin{bmatrix} 29 & -9 \\ -9 & 3 \end{bmatrix}\begin{bmatrix} 21 \\ 67 \end{bmatrix} = \frac{1}{6}\begin{bmatrix} 609 - 603 \\ -189 + 201 \end{bmatrix} = \begin{bmatrix} 1 \\ 2 \end{bmatrix}$$

So: $\hat{y} = 1 + 2x$ ✓

## **When OLS Closed-Form Fails**

The formula requires $(\mathbf{X}^T\mathbf{X})^{-1}$ to exist. This **fails** when:

1. **Perfect multicollinearity**: Two predictors are perfectly correlated (e.g., $x_2 = 2x_1$)
2. **$n < p+1$**: More parameters than observations (matrix is singular)
3. **Numerical instability**: $(\mathbf{X}^T\mathbf{X})$ is ill-conditioned

**Solutions:** Regularization (Ridge: add $\lambda\mathbf{I}$), drop variables, or use SVD/pseudoinverse.
