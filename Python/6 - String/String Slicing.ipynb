```python
# String Slicing
print("String Slicing")
str = "Jack Sparrow"
print("Length of str : ", len(str))
print(str[0:100])
print(str[:])
print(str[:4])
print(str[2:7])
print(str[2: -1]) # -1 represents -> len(str) - 1; 
print(str[2: len(str)-1])
print(str[-4: -1])
print(str[len(str)-4: len(str)-1])
print(str[len(str)-1: len(str)-4]) # no error but Doen not print enything
```
